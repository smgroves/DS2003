### Redesign checklist

These are the same criteria from Wednesday's handout. Before you turn in your lab, check your final chart against each one.

**Question and task**
- [ ] Your chart answers one clear, specific question, and a reader could figure out that question without you explaining it.
- [ ] The chart type fits the primary task (e.g., comparing values, finding a trend, looking up a value, seeing a distribution) and the attribute types you have.

**Encoding: expressiveness and effectiveness**
- [ ] **Expressiveness:** each channel matches its attribute type (nominal, ordinal, interval, ratio). Don't imply order where there is none, and don't hide order where there is some.
- [ ] **Effectiveness:** the most important attribute is on the most accurate channel (position on a common scale > length > angle/area > color).
- [ ] **Accuracy / lie factor:** the size of a visual effect matches the size of the effect in the data.

**Scales and axes**
- [ ] **Scale choices:** bars start at zero, a log scale is used only when it's justified and is clearly labeled, and there are no hidden or unexplained scale breaks. If you use two y-axes, you can defend why.
- [ ] **Aspect ratio:** the shape of the plot doesn't exaggerate or flatten trends (bank lines toward 45°).
- [ ] **Domain to range mapping:** axis limits and tick marks fit the data and are easy to read.

**Perception**
- [ ] **Use of color:** color has a purpose, the palette type matches the data (categorical, sequential, or diverging), and the chart is readable by people with color vision deficiency. Color isn't the only way to tell groups apart.
- [ ] **Weber's law:** the differences you want readers to see are large relative to the baseline, or you've added gridlines or reference lines to help.
- [ ] **Stevens' power law:** you avoid area and volume for values readers need to judge precisely.
- [ ] **Discriminability:** readers can tell the categories or color levels apart (keep to a small number of hues).
- [ ] **Separability:** combined channels don't interfere with each other (e.g., size and color on the same mark).

**Gestalt and emphasis**
- [ ] You use **proximity, similarity, symmetry, connectedness, continuity, closure, common fate,** and/or **pop-out** to make the main message obvious.

**Chart anatomy**
- [ ] An informative title (ideally one that states the takeaway), axis labels with units, readable tick labels, and a legend or direct labels.
- [ ] Annotations where they help (e.g., marking an event), and a source or caption line.

**Process**
- [ ] Your redesign incorporates the feedback from **both** groups that critiqued your sketch.
- [ ] If your chart comes with a context dataset, your final chart uses information from **both** datasets.
