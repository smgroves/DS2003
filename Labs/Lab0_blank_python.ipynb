**Python's design**

- General-purpose language, not built specifically for statistics — but its data libraries (`pandas`, `numpy`) give it strong statistical/tabular capabilities
- Huge community and ecosystem, especially for data science
- Dynamically typed: a variable's type is decided at runtime, not declared up front
- Indentation (not braces) defines blocks of code
- Almost everything is an object, including functions and modules themselves

**Python syntax basics**

- Blocks are formed by **indentation**, not `{ }` — this is not optional, it's part of the syntax
- Statements normally end at the newline; semicolons are optional and rarely used
- Assignment uses `=`
- Comments start with `#`
- Single and double quotes for strings are interchangeable; pick one and be consistent (use the other kind to embed quotes without escaping)

**List indexing**

Retrieve values with square brackets `[ ]`.

**Python indexes are 0-based** — the first element is index `0`, not `1`.

```python
# retrieve the 3rd value (remember: 0-based!)
s = ["aa", "bb", "cc", "dd", "ee"]

```

---
**Negative indexing**

A negative index counts backward from the *end* of the list — `s[-1]` is the last element, `s[-2]` is second-to-last, and so on.

```python
# example
s
s[-1]   # last element
```

### Out-of-range indexes

An out-of-range index raises an `IndexError`. Try it and read the error message.

```python
# example — this will raise an error, that's expected
s[10]
```

---
**Slicing with multiple indices**

Python doesn't let you index a plain list with an arbitrary list of positions directly — but you can get the same effect with a list comprehension, or by converting to a `numpy` array / `pandas` Series (which support this directly, called "fancy indexing").

```python
s = ["aa", "bb", "cc", "dd", "ee"]

# retrieve the 2nd and 3rd elements (0-based: indices 1 and 2) using a list comprehension
[s[i] for i in [1, 2]]
```

**Duplicate and out-of-order selections** work the same way with a list comprehension:

```python
[s[i] for i in [1, 2, 2]]
```

```python
[s[i] for i in [1, 0, 2]]
```

**Range slicing**

Python's slice syntax `s[start:stop]` retrieves a range directly — no separate function needed, and the `stop` index is *excluded*.

```python
s[1:4]
```

## Dictionaries

Python's `dict` is a collection of key/value pairs.


```python
n = [2, 3, 5]
s = ["aa", "bb", "cc", "dd", "ee"]
b = [True, False, True, False, False]

x = {"numbers": n, "strings": s, "flags": b, "extra": 3}   # a dict containing n, s, b, and the number 3
x
```

Each dict entry is accessed by its **key**, not by position — this is the main difference from a list.

### Accessing dict entries

Retrieve a value with square brackets and the key name.

```python
# retrieve the value stored under the key "strings" — a copy of s
x["strings"]
```

---
With a list of keys, you can retrieve several entries at once using a small loop or dict/list comprehension — there's no single-shot multi-key indexing built into `dict`.

```python
# retrieve the "strings" and "extra" entries together
[x[k] for k in ["strings", "extra"]]
```

## Conditional Statements

Run code based on a condition with `if`:

```python
if True:
    print("This will print.")
```

`if`, `elif`, and `else` chain together for multi-way conditional logic (note: Python uses `elif`, not `else if`).

```python
x = 10

if x > 10:
    print("x is greater than 10")
elif x == 10:
    print("x is exactly 10")
else:
    print("x is less than 10")
# Output: "x is exactly 10"
```

## Loops

Loops iterate over sequences. `range(1, 6)` generates 1 through 5 — remember the stop value is excluded.

```python
for i in range(1, 6):
    print(i)
```
